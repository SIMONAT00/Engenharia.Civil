CUSTO DE MATERIAIS PARA SUA OBRA

```python
# Indicar o que será pedido nesse comando
import sys
texto1 = "Este orçamento tem como objetivo apresentar a estimativa de custo para a execução de um projeto,"
print(texto1)
texto2 = "levando em consideração os materiais. A seguir, detalharemos os valores para a realização de sua obra. para isso precisamos de informações. "
print(texto2)
# Perguntar se ele aceita nossa ajuda
while True:
  resposta = input("Você deseja continuar? (sim/não): ").lower()
  if resposta == "sim":
    break
    print("Otimo! podemos fazer algumas perguntas?.")
  elif resposta == "não":
    print("Tudo bem, agradecemos.")
    sys.exit()
  else:
    print("Resposta inválida. Por favor, digite 'sim' ou 'não'.")
# caso ele responda não encerrar
# Caso continue Perguntar nome e endereço do cliente
cli = input("Nome do cliente: ")
end = input("Endereço da obra: ")
# Explicar média de quantidade e custo por material
texto3 = "A média de materias utilazados por metro quadrado são:"
print(texto3)
texto4 = "Areia em média 1,4 por m², com custo médio de 68,75R$ "
print(texto4)
texto5 = "Brita em média 1,43 por m², com custo em médio de 199,40R$ "
print(texto5)
texto6 = "Cimento em média 0,5 saco por m, com custo em médio de 25,90R$² "
print(texto6)
texto7 = "Blocos de cimento em média 14 unidades por m², com custo em médio de 3,48R$ "
print(texto7)
texto7 = "Ferro três quartos em média 11,75 KG por m², com custo em médio de 77,41R$ "
print(texto7)
# Pedir para o cliente digitar o tamanho da obra que dejesa contruir
fre = float(input("Indique o comprimento do que deseja contruir em m: "))
fun = float(input("Indique a largura do que deseja contruir em m: "))
print(f"Então {cli} logo a seguir seu custo médio estará pronto:" )

# Comando para agurdar um instante
import time
tempo = 5
for i in range(tempo, 0, -1):
    print(i)
    time.sleep(1)


# mostrar custo pra o cliente
texto8 = "PRONTO ESTÁ AQUI SEU ORÇAMENTO !!:"
print(texto8)
m = fre * fun
print(f"Sua construção terá o tamanho de: { m } m² localizado em {end}" )
areia = 69 * 2
cst1 = m * areia
print(f"voce ira gastar em média com areia: { cst1 } R$" )
brita = 200 * 2
cst2 = m * brita
print(f"voce ira gastar em média com brita: { cst2 } R$" )
cimento = 26 * 1
cst3 = m * cimento
print(f"voce ira gastar em média com cimento: { cst3 } R$" )
bc = 14 * 4
cst4 = m * bc
print(f"voce ira gastar em média com bloco de cimento: { cst4 } R$" )
ftq = 78 * 12
cst5 = m * ftq
print(f"voce ira gastar em média com ferro três quartos : { cst5 } R$" )

# Indicar soma de todos os materiais
soma = cst1 + cst2 + cst3 + cst4 + cst5
print(f"Seu custo total será de: {soma} R$")

# Indicar média de materiais por metro quadrado
media = soma/m
print(f"A média de custo por metro quadrado é de: {media} R$")
if  media <= 1556:
  # Indicar se esta tudo certo e se o cliente pode costruir
  print("Está tudo em ordem sua obra está na média")
else:
  print("Sua obra está acima da média")
```
